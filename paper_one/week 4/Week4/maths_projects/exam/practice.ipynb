```python
import matplotlib.pyplot as plot

x = [1, 2, 3, 4, 5]
y = [3, 6, 9, 12, 15]

plot.plot(x, y, marker = "o")
plot.show()
```

```python
import numpy as np

A = [70, 72, 68, 71, 69]
B = [40, 100, 50, 90, 70]

sum_A = np.sum(A)
sum_B = np.sum(B)

print(sum_A, sum_B)

variance_A = np.var(A)
variance_B = np.var(B)

print(variance_A, variance_B)
```
