```python
def is_even(number):
    if number % 2 == 0:
        return True
    else:
        return False
    
is_even(10)    
```

```python
def is_vowels(word):
    check_this_letters = ["a", "e", "i", "o", "u"]
    
    count = 0
    
    lower_letter_word = word.lower()
    
    for letter in lower_letter_word:
        if letter in check_this_letters:
            count += 1
            
    return count

is_vowels("Abi")
```
