```python
import pandas as pd

data = {
    "Name" : ["A", "B", "C", "D", "E"],
    "Age" : [10, 20, 15, 18, None],
    "City" : ["Tvm", "chennai", "Kollam", "Tvm", "Kochi"]
}

info = pd.DataFrame(data)

info.index += 1
print(info)
info.info()
info.isnull().sum()
info.dropna()
info.dropna(subset= "Name")
```

```python
print(info.shape)
print(info.dtypes)
print(info["Age"].mean())
info["Age"] = info["Age"].fillna(info["Age"].mean())
info["Age"].astype(int)
print(info)
```
