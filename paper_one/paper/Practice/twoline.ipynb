```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]

fig, axis = plt.subplots()

axis.plot(x, y1, label = 'line1', marker = 'o')
axis.plot(x, y2, label = 'line2', marker = 's')

axis.set_xlabel("X-Axis")
axis.set_ylabel("Y-Axis")

axis.legend()
plt.show()
```
