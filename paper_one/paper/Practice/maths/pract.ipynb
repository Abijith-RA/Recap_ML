```python
def mean(values):
    
    lenght = len(values)
    
    total = 0
    
    for num in values:
        total += num
        
    mean = total/lenght
    
    return mean

A = [1, 10, 20, 9, 10, 10]

mean(A)
```

```python
def meadian(numbers):
    
    lenght = len(numbers)
    mid = lenght // 2
    
    if lenght % 2 == 1:
    
        meadian_number = numbers[mid]
            
        return meadian_number
    
    else:
        
        meadian_number = (numbers[mid-1] + numbers[mid]) / 2
        
        return meadian_number
        

meadian(A)
    
```

```python
def mode(numbers):
    
    maxcount = 0
    mode_value = None
    
    for i in numbers:
        
        count = 0
        
        for j in numbers:
            if i == j:
                count += 1
                
        if count > maxcount:
            maxcount = count
            mode_value = i
            
    return mode_value

mode(A)
        
```

```python
def varience(numbers):
    
    mean_of_numbers = mean(numbers)
    lenght = len(numbers)
    sumation = 0
    
    for num in numbers:
        
        sumation += (num - mean_of_numbers) ** 2
        varience_numbers = sumation / lenght
        
    return varience_numbers

varience(A)
```

```python
def standardvarience(numbers):
    
    mean = sum(numbers) / len(numbers)
    summation = 0
    
    for num in numbers:
        summation += (num - mean) ** 2
        varience_before = summation / len(numbers)
        
    return varience_before ** 0.5

standardvarience(A)
```

```python
def calculator(a, b, operation):
    try:
        if operation == "+":
            result = a + b
        elif operation == "-":
            result = a - b
        elif operation == "*":
            result = a * b
        elif operation == "/":
            result = a / b
        else:
            raise ValueError("Invalid operation")
    
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
        result = None
    
    except ValueError as e:
        print(f"Error: {e}")
        result = None
    
    else:
        print("Calculation successful")
    
    finally:
        print("Calculator finished running")
    
    return result


print(calculator(10, 5, "/"))
print("---")
print(calculator(10, 0, "/"))
print("---")
print(calculator(10, 5, "%"))
```

```python
def fib(number):
    a = 0
    b = 1
    
    for i  in range(number):
        print(a, end=" ")
        
        a, b = b, a + b

fib(7)    
```

```python
def prime(number):
    
    if number < 2:
        print("Not a prime")
    else:
        is_prime = True
    
    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break
    
    if(is_prime):
        print("Prime number")
    else:
        print("Not a prime") 
        
prime(3)           
```

```python
def dot_product(v1, v2):
    
    result = 0
    
    for i in range(len(v1)):
        result += v1[i] * v2[i]
        
    return result

vector_one = [1, 2, 3, 4, 5]
vector_two = [1, 1, 1, 1, 1]

dot_product(vector_one, vector_two)

import matplotlib.pyplot as plt

plt.plot(vector_one,vector_two)
```

```python
def matric_multi(m1, m2):
    
    
    
    row1 = len(m1)
    colum1 = len(m1[0])

    row2 = len(m2)
    colum2 = len(m2[0])
    
    result = [[0] * colum2 for i in range(row1)]
    
    for i in range(row1):#0
        for j in range(colum2):#01
            for k in range(row2):#0101
                result[i][j] += m1[i][k] * m2[k][j]
                
    return result            

matric_one = [[1, 2, 3],
              [4, 5, 6]]
matric_two = [[9, 8],
              [7, 6],
              [5, 4]]    

matric_multi(matric_one,matric_two)
```
