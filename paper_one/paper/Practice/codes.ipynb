```python
import random


def gradeCalculator(marks):
    
    print(marks)
    
    Grade = ["A" if ms > 90 else "B" if ms > 80 else "C" if ms > 70 else "D" if ms > 60 else "E" if ms > 50 else "Fail" for ms in marks]
    
    return Grade


scores = [random.randint(10, 100) for _ in range(10)]

gradeCalculator(scores)
```
