Write a function scalar_ops(a, b) that takes two scalars and returns their sum, difference, product, and quotient as a tuple.

```python

def scalar_ops(a, b):
    
    sum = a + b
    
    difference = a - b
    
    product = a * b
    
    quotient = a / b

    
    return (sum, difference, product, quotient)

scalar_ops(10, 3)
```

Write a function vector_scalar_multiply(v, k) that multiplies every element of a vector by scalar k.

```python
def vector_scalar_multiply(v, k):
    
    result = []
    
    for i in range(len(v)):
        
        value = v[i] * k
        
        result.append(value)
        
    return result

vector_scalar_multiply([2, 4, 6], 3)    
```

Write same_direction(v1, v2, tolerance=1) that returns True if two vectors' angles are within tolerance degrees of each other, otherwise False.

```python
import math

def same_direction(v1, v2, tolerance = 1):
    
    degree_one = math.degrees(math.atan2(v1[1], v1[0]))
    
    degree_two = math.degrees(math.atan2(v2[1], v2[0]))
    
    diffrence = abs(degree_one - degree_two)
    
    if diffrence <= tolerance:
        return True
    else:
        return False

# vector_one = [0, 4]
# vector_two = [1, 2]

same_direction([1, 1], [2, 2])
```

Write dot_product(v1, v2) from scratch with a length-check. Then use it to write angle_between(v1, v2) that returns the actual angle (in degrees) between two vectors using the formula: angle = arccos(dot_product(v1,v2) / (magnitude(v1) * magnitude(v2))).

```python
import math

def dot_product(v1, v2):
    
    if(len(v1) != len(v2)):
        
        return("the vectors not same size")
    else:
        
        result = 0
        
        for i in range(len(v1)):
            
            result += v1[i] * v2[i]
            
        return result
    
def magnitude(v):
    
    total = 0
    
    for num in v:
        
        total += num ** 2
        
    result = total ** 0.5
    
    return result

def angle_between(v1, v2):
    
    dotResult = dot_product(v1, v2)
    magV1 = magnitude(v1)
    magV2 = magnitude(v2)
    
    toCos = (dotResult / (magV1 * magV2))
    
    angle_radius = math.acos(toCos)
    degree_angle = math.degrees(angle_radius)
    
    return degree_angle

        
vector_one = [10, 2]
vector_two = [20, 6]

# print(magnitude(vector_one))
# print(magnitude(vector_two))

print(dot_product(vector_one,vector_two))

# print(angle_between(vector_one, vector_two))
```

```python
# Simple demonstration: using the derivative to take one "learning step"
def f(x):          # imagine this is a simplified "loss" function
    return (x - 5) ** 2

def derivative_at_point(f, x, h=0.0001):
    return (f(x + h) - f(x)) / h

x = 0                     # starting guess (like a starting weight)
learning_rate = 0.1

for step in range(10):
    grad = derivative_at_point(f, x)
    x = x - learning_rate * grad    # move AGAINST the slope, toward the minimum
    print(f"Step {step}: x = {x:.3f}, loss = {f(x):.3f}")

# x should get closer and closer to 5 (the true minimum) with each step
```
