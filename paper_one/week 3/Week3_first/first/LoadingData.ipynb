```python
import pandas as pd

# Create DataFrame
data = {
    "Age": [22, 25, 30, 28, 35, 40],
    "Salary": [30000, 40000, 50000, 45000, 60000, 70000]
}

df = pd.DataFrame(data)

# Histogram using Pandas
df["Age"].hist()

# Scatter plot using Pandas
df.plot(kind="scatter", x="Age", y="Salary")
```

```python
df[df["Age"] > 100]          # find rows where age is unrealistically high
df[df["Salary"] < 0]           # find rows with impossible negative salary
```
