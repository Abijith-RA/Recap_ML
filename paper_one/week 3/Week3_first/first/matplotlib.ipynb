```python
import matplotlib.pyplot as plt

ages = [22,25,28,30,35,40,45,50]
salaries = [30000,35000,40000,42000,50000,55000,60000,62000]

plt.scatter(ages, salaries, color="purple", alpha=0.6)
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [100, 120, 115, 140, 160]

plt.plot(days, sales, marker="o", color="steelblue", label="Sales")
plt.title("Sales Trend Over 5 Days")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()  # Add legend
plt.show()
```

```python
ages = [22,25,25,28,30,30,30,35,40,45,45,50]

plt.hist(
    ages,
    bins=5,
    color="mediumseagreen",
    edgecolor="black",
    alpha=0.7,
    linewidth=3,
    label="Age Distribution",
    density=False
)
plt.legend()
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
import matplotlib.pyplot as plt

ages = [18, 20, 21, 22, 25, 25, 26, 30, 32, 35, 35, 40, 45]

plt.hist(ages, bins=5, color="mediumseagreen", edgecolor="black", density=False)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution (Count)")
plt.show()
```

```python
import matplotlib.pyplot as plt

ages = [18, 20, 21, 22, 25, 25, 26, 30, 32, 35, 35, 40, 45]

plt.hist(ages, bins=5, color="tomato", edgecolor="black", density=True)

plt.xlabel("Age")
plt.ylabel("Density")
plt.title("Age Distribution (Probability Density)")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create age data
df = pd.DataFrame({
    "Age": [18, 20, 22, 25, 25, 28, 30, 32, 35, 40, 45, 50],
    "City": ["NY", "NY", "NY", "London", "London", "London",
             "Tokyo", "Tokyo", "Tokyo", "NY", "London", "Tokyo"],
    "Salary": [60000, 65000, 70000, 45000, 50000, 55000,
               70000, 75000, 80000, 72000, 52000, 85000]
})

# Density plot for Age
df["Age"].plot(kind="density", color="green")

plt.title("Age Distribution (Density Plot)")
plt.xlabel("Age")
plt.ylabel("Density")

plt.show()


# Box plot for Salary by City
df.boxplot(column="Salary", by="City")

plt.title("Salary Distribution by City")
plt.suptitle("")  # removes automatic title
plt.xlabel("City")
plt.ylabel("Salary")

plt.show()
```
