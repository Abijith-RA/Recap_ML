```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import joblib

data = pd.DataFrame({
    "experience": [1, 2, 3, 4, 5, 6],
    "salary": [20000, 30000, 40000, 50000, 60000, 70000]
})

X = data[["experience"]]
y = data["salary"]

model = LinearRegression()
model.fit(X, y)

joblib.dump(model, "salary_prediction_model.joblib")
```
