```python
#age mean

data = {
    "Name": ["Arun", "Meera", "Rahul", "Anu"],
    "Age": [25, None, 30, 28],
    "Salary": [30000, 40000, None, 35000]
}
```

```python
import pandas as pd

df = pd.DataFrame(data)
```

```python
df.info()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df["Salary"] = df["Age"].fillna(df["Age"].median())
```
