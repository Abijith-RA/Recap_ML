```python
df = df.dropna(subset=["salary"])
df.info()
```

```python
from sklearn.model_selection import train_test_split

x = df[["age", "city", "department", "join_date"]]
y = df["salary"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
x_train.info()
```

```python
x_train.info()
```

```python
age_mean = df["age"].mean()
```

```python
x_train["age"] = x_train["age"].fillna(age_mean)
x_test["age"] =x_test["age"].fillna(age_mean)
```

```python
x_train["city"] = x_train["city"].fillna(x_train["city"].mode()[0])
x_test["city"] = x_test["city"].fillna(x_train["city"].mode()[0])
```

```python
x_train["department"] = x_train["department"].fillna(x_train["department"].mode()[0])
x_test["department"] = x_test["department"].fillna(x_train["department"].mode()[0])
```

```python
x_train["join_date"] = x_train["join_date"].fillna(x_train["join_date"].mode()[0])
x_test["join_date"] = x_test["join_date"].fillna(x_train["join_date"].mode()[0])
```

```python
x_train.info()
x_test.info()
```

```python
x_train = pd.get_dummies(x_train,columns=["city", "department"], dtype=int)
```

```python
x_test = pd.get_dummies(x_test, columns=["city", "department"], dtype=int)
x_test = x_test.reindex(columns=x_train.columns, fill_value=0)
```

```python
x_train["join_date"] = pd.to_datetime(x_train["join_date"])
```

```python
x_train["join_year"] = x_train["join_date"].dt.year
x_train["join_month"] = x_train["join_date"].dt.month
x_train["join_day"] = x_train["join_date"].dt.day
```

```python
x_train.to_csv("ready_date", index= False)
```

```python
x_test["join_date"] = pd.to_datetime(x_test["join_date"])
```

```python
x_test["join_year"] = x_test["join_date"].dt.year
x_test["join_month"] = x_test["join_date"].dt.month
x_test["join_day"] = x_test["join_date"].dt.day
```

```python
x_train = x_train.drop(columns=["join_date"])
x_test = x_test.drop(columns=["join_date"])
```

```python
x_test.to_csv("test_readyData", index=False)
```

```python
y_train.to_csv("output_to_train", index=False)
```

```python
x_test.to_csv("xtest_data", index=False)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train,y_train)

y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = root_mean_squared_error(y_test, y_pred)   # no more squared= argument
r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R2  :", r2)
```
