```python
data = {
    "Name" : ["abi", "achu", "abijith", None],
    "Age" : [1, 22, 27, 23],
    "salary" : ["$2100", 2000, "$2500,50", 500]
}

import pandas as pd

df = pd.DataFrame(data)
```

```python
df
```

```python
df = df.dropna(subset=["Name"])
```

```python
df
```

```python
df.loc[df["Age"] < 18,"Age"] = df["Age"].mean().astype(int)
```

```python
df
```

```python
df.info()
```

```python
df["salary"] = df["salary"].astype(str).str.replace(r'[^\d,]', "", regex=True)
```

```python
df["salary"] = df["salary"].astype(str).str.replace(",",".", regex=False).astype(float).astype(int)
```
