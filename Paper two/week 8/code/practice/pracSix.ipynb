```python
df["salary"] = df["salary"].astype(str).str.replace(r'[^\d.]','',regex=True).astype(float)
```

```python
columns_to_check = ["department", "city", "education", "gender", "remote_work", "performance_rating"]

for col in columns_to_check:
    print(f"\n--- {col} ---")
    print(df.groupby(col)["salary"].mean().sort_values(ascending=False))
```
