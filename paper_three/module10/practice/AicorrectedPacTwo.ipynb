```python
# %%
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("/home/thinkpad/Documents/BrototypeStudying/paper_three/module10/data/employees_practice.csv")

# %%
df.shape

# %%
df.info()

# %%
df.isnull().sum()

# %%
df = df.drop_duplicates()

# %%
df.dropna(subset=["salary"], inplace=True)

# %%
corr_matrix = df.corr(numeric_only=True)
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")

# %%
df = df.drop(columns=["employee_id", "name", "rating"])

# %%
df["department"] = df["department"].str.strip().str.lower()
df = pd.get_dummies(df, columns=["department"], dtype=int)

# %%
from sklearn.model_selection import train_test_split

x = df.drop(columns=["salary"])
y = df["salary"]

x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=.80, random_state=40)

# %%
# fill missing values using TRAIN medians only
for col in ["age", "experience_years"]:
    median = x_train[col].median()
    x_train[col] = x_train[col].fillna(median)
    x_test[col] = x_test[col].fillna(median)

# %%
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
cols = ["age", "experience_years"]

x_train[cols] = scaler.fit_transform(x_train[cols])
x_test[cols] = scaler.transform(x_test[cols])

# %%
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)
prediction = model.predict(x_test)

# %%
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print(f"MAE : {mean_absolute_error(y_test, prediction):.2f}")
print(f"MSE : {mean_squared_error(y_test, prediction):.2f}")
print(f"R^2 : {r2_score(y_test, prediction):.2f}")
```
