```python
import numpy as np
import plotly.graph_objects as go
from sklearn.linear_model import LinearRegression

# Sample data: house size, rooms -> price
rng = np.random.default_rng(0)
size = rng.uniform(500, 3000, 40)
rooms = rng.uniform(1, 6, 40)
price = 50 + 0.12 * size + 15 * rooms + rng.normal(0, 20, 40)

# Fit multiple linear regression
X = np.column_stack([size, rooms])
model = LinearRegression().fit(X, price)

# Build the best-fit plane
xx, yy = np.meshgrid(
    np.linspace(size.min(), size.max(), 20),
    np.linspace(rooms.min(), rooms.max(), 20),
)
zz = model.intercept_ + model.coef_[0] * xx + model.coef_[1] * yy

# Plot points + plane
fig = go.Figure([
    go.Scatter3d(x=size, y=rooms, z=price, mode="markers",
                 marker=dict(size=4, color="purple"), name="Houses"),
    go.Surface(x=xx, y=yy, z=zz, opacity=0.5,
               colorscale="Teal", showscale=False, name="Best-fit plane"),
])
fig.update_layout(scene=dict(xaxis_title="Size",
                             yaxis_title="Rooms",
                             zaxis_title="Price"))
fig.show(renderer="browser")
```
