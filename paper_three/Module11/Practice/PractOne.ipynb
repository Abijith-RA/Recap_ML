```python
import plotly.graph_objects as go

# Linear data
temp_linear = [20, 25, 30]
sales_linear = [100, 150, 200]

# Non-linear data
temp_nonlinear = [20, 25, 30, 35]
sales_nonlinear = [100, 120, 200, 400]

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=temp_linear, y=sales_linear,
    mode="lines+markers", name="Linear (straight line)",
    line=dict(color="royalblue", width=3)
))

fig.add_trace(go.Scatter(
    x=temp_nonlinear, y=sales_nonlinear,
    mode="lines+markers", name="Non-linear (curve)",
    line=dict(color="crimson", width=3)
))

fig.update_layout(
    title="Temperature vs Ice Cream Sales",
    xaxis_title="Temperature (°C)",
    yaxis_title="Ice Cream Sales",
    template="plotly_white"
)

fig.show(renderer="browser")
```
