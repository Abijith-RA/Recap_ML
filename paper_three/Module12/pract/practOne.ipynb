```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

df = pd.read_csv("/home/abijithpersonal/Documents/Recap_ML/paper_three/Module12/data/customer_segmentation.csv")
features = df.select_dtypes("number").drop(columns=["customer_id"])

X_scaled = StandardScaler().fit_transform(features)
model = KMeans(n_clusters=2, random_state=42, n_init=10)
model.fit(X_scaled)

plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=model.labels_)
plt.xlabel(features.columns[0])
plt.ylabel(features.columns[1])
plt.show()
```
