```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

df = pd.read_csv("/home/thinkpad/Documents/BrototypeStudying/paper_three/Module12/data/customer_segmentation.csv")

X = df.drop(columns="customer_id")

model = KMeans(n_clusters=3, n_init=10, random_state=42)
model.fit(X)

raw_label = model.predict(X)

print(silhouette_score(X, raw_label))
```

```python
scaler = StandardScaler()

x_scaled = scaler.fit_transform(X)

model.fit(x_scaled)

scaled_pred = model.predict(x_scaled)

print(silhouette_score(x_scaled, scaled_pred))
```
