```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv("/home/thinkpad/Documents/BrototypeStudying/paper_three/Module12/data/customer_segmentation.csv")
X = df.select_dtypes("number").drop(columns=["customer_id"])
X_scaled = StandardScaler().fit_transform(X)

pca = PCA().fit(X_scaled)
print(pca.explained_variance_ratio_)
print(np.cumsum(pca.explained_variance_ratio_))

pca95 = PCA(n_components=0.95).fit(X_scaled)   # keep enough PCs for 95% variance
print(pca95.n_components_)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

df = pd.read_csv("/home/thinkpad/Documents/BrototypeStudying/paper_three/Module12/data/customer_segmentation.csv")
X = df.select_dtypes("number").drop(columns=["customer_id"])
X_scaled = StandardScaler().fit_transform(X)
X_pca = PCA(n_components=2).fit_transform(X_scaled)

full = KMeans(n_clusters=3, random_state=42, n_init=10).fit_predict(X_scaled)
small = KMeans(n_clusters=3, random_state=42, n_init=10).fit_predict(X_pca)
print("all columns:", silhouette_score(X_scaled, full))
print("2 PCs:", silhouette_score(X_pca, small))
```
