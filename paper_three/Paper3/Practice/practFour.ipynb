```python
import pandas as pd

df = pd.read_csv("/home/thinkpad/Documents/BrototypeStudying/paper_three/Paper3/data/classification_students.csv")
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns="passed")
y = df["passed"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=.20, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import KFold, cross_val_score, GridSearchCV

kf = KFold(n_splits=5, shuffle=True, random_state=42)

params = {"n_estimators" : [50, 100, 150],
          "max_depth" : [3, 5, 10]}

grid = GridSearchCV(RandomForestClassifier(random_state=42), params, cv=kf, scoring="accuracy")

grid.fit(x_train, y_train)
print(grid.best_params_, grid.best_estimator_)
```

```python
bestModel = grid.best_estimator_
print(bestModel.score(x_test, y_test))
```

```python
from sklearn.model_selection import RandomizedSearchCV

rand = RandomizedSearchCV(RandomForestClassifier(random_state=42), params, n_iter=5, cv=kf, scoring="accuracy", random_state=42)

rand.fit(x_train, y_train)
print(rand.best_estimator_, rand.best_score_)
```

```python
bestModel = rand.best_estimator_
print(bestModel.score(x_test, y_test))
```

```python
from sklearn.tree import DecisionTreeClassifier, export_text

model = DecisionTreeClassifier(max_depth=2, random_state=42)
model.fit(x_train, y_train)                    # the tree must be trained first

print(export_text(model, feature_names=list(x_train.columns)))
```
