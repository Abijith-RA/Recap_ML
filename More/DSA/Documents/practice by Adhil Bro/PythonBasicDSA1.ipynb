```python
#FizzBuzz problem
class solution:
    def FizzBuzz(self,n:int)->list[int]:
        ans=[]
        for i in range(1,n+1):
            if i%3==0 and i%5==0:
                ans.append('FizzBuzz')
            elif i%3==0:
                ans.append('Fizz')
            elif i%5==0:
                ans.append('Buzz')
            else:
                ans.append(str(i))
        return ans
obj=solution()
result=obj.FizzBuzz(15)
print(f'Result is:{result}')
```

```python
#Count Odd number in an interval range
class solution:
    def CountOddNumberInRange(self,low:int,high:int)->int:
        count=0
        for i in range(low,high+1):
            if i%2!=0:
                count+=1
        return count
obj=solution()
result=obj.CountOddNumberInRange(2,6)
print(f'Result is:{result}')
```

```python
#How many numbers smaller than the current number
#"Should this variable remember its old value when the next outer-loop iteration starts?"
class solution:
    def SmallerThanCurrent(self,nums:list[int])->list[int]:
        ans=[]
        for i in nums:
            count=0
            for j in nums:
                if j<i:
                    count+=1
            ans.append(count)
        return ans
obj=solution()
result=obj.SmallerThanCurrent([8,1,2,2,3])
print(f'Result is:{result}')
```

```python
#Count digit that divides number
#NOTE: We use a temperary variable to store number , n → kept unchanged so we can check whether the original number is divisible by each digit.
class solution:
    def CountDigits(self,n:int)->int:
        temp=n
        count=0
        while temp>0:
            rem=temp%10
            if n%rem==0:
                count+=1
            temp//=10
        return count
obj=solution()
result=obj.CountDigits(111)
print(f"Result is:{result}")
```

```python
#Palindrome
class solution:
    def Palindrome(self,n:int)->bool:
        temp=n
        rev=0
        while temp>0:
            rem=temp%10
            rev=rev*10+rem
            temp//=10
        if n==rev:
            return True
        else:
            return False
obj=solution()
result=obj.Palindrome(1991)
print(f'The Given number is palindrome:{result}')
```

```python
#Subtracting the product and sum of number
class solution:
    def SubtractProductandSum(self,n:int)->int:
        temp=n
        sum=0
        product=1
        while temp>0:
            rem=temp%10

            sum+=rem
            product*=rem

            temp//=10

        return product-sum
obj=solution()
result=obj.SubtractProductandSum(234)
print(f'Result is:{result}')
```
