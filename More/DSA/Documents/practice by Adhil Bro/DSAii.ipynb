```python
#Every time we are compring adjacent elements.
#Space complexity O(1)
#Time complexity a)Best Case: O(n**2)
#                b)Worst Case:O(n)
class solution:
    def BubbleSort(self,arr):
        n=len(arr)
        isSwap=False
        for i in range(n-1):
            for j in range(n-i-1):
                if arr[j]>arr[j+1]:
                    temp=arr[j]
                    arr[j]=arr[j+1]
                    arr[j+1]=temp
                    isSwap=True
            if not isSwap:
                break
        return arr
obj=solution()
r=obj.BubbleSort([5,1,4,2,8])
print(f'Result is:{r}')
```

```python
#Shifting elements if the condition is true
#Space complexity:O(1)
#TIme complexity: worst case->O(n**2)
#                   best case->O(n)
class solution:
    def Insertionsort(self,arr):
        n=len(arr)
        for i in range(1,n):
            key=arr[i]
            j=i-1

            while j>=0 and arr[j]>key:
                arr[j+1]=arr[j]
                j-=1
            arr[j+1]=key
        return arr
obj=solution()
r=obj.Insertionsort([1,2,5,7,9,4])
print(f'Result is:{r}')
```

```python
#First find minimum element then swap
#Space complexity:O(1)
#TIme complexity: worst case->O(n**2)
#                   best case->O(n**2)
class solution:
    def SelectionSort(self,arr):
        n=len(arr)
        for i in range(n):
            min_index=i

            for j in range(i+1,n):
                if arr[j]<arr[min_index]:
                    min_index=j
            temp=arr[i]
            arr[i]=arr[min_index]
            arr[min_index]=temp
        return arr
        
obj=solution()
r=obj.SelectionSort([7,3,5,2,9])
print(f'Result is:{r}')
            
```

```python
#Merge sort is a recursive algorithm
class solution:
    def Merge(self,nums,l,mid,r):
        a=[]
        b=[]

        for i in range(l,mid+1):
            a.append(nums[i])
        for i in range(mid+1,r+1):
            b.append(nums[i])

        i,j,k=0,0,l

        while k<=r:
            if j==len(b):
                nums[k]=a[i]
                i+=1
                k+=1
            elif i==len(a):
                nums[k]=b[j]
                j+=1
                k+=1
            elif a[i]<b[j]:  
                nums[k]=a[i]
                i+=1
                k+=1
            else:
                nums[k]=b[j]
                j+=1
                k+=1
    def MergeSort(self,nums,l,r):
        #Base Case
        if l>=r:
            return
        #Recursive case
        mid=(l+r)//2
        self.MergeSort(nums,l,mid)
        self.MergeSort(nums,mid+1,r)

        self.Merge(nums,l,mid,r)

    def SortArray(self,nums):
        self.MergeSort(nums,0,len(nums)-1)
        return nums

obj=solution()
r=obj.SortArray([5,2,3,1,4])
print(f"Result is:{r}")
```

```python
class solution:
    def Partition(self,nums,l,r):
        key=nums[r]
        start=l

        for i in range(l,r+1):
            if nums[i]<=key:
                temp=nums[i]
                nums[i]=nums[start]
                nums[start]=temp
                start+=1
        return start-1
    def QuickSort(self,nums,l,r):
        #Base Case
        if l>=r:
            return
        
        #Recursive Case
        p=self.Partition(nums,l,r) 

        self.QuickSort(nums,l,p-1)
        self.QuickSort(nums,p+1,r)
        
    def SortArray(self,nums):
        n=len(nums)
        self.QuickSort(nums,0,n-1)

        return nums

obj=solution()
r=obj.SortArray([5,2,3,1,4])
print(f"Result is:{r}")
```

```python
class HashTable:
    def __init__(self,size):
        self.size=size
        self.table=[[] for _  in range(size)]

    def hashfunction(self,key):
        return key%self.size
    
    def insert(self,key):
        index=self.hashfunction(key)
        if key not in self.table[index]:
            self.table[index].append(key)
            print(f'{key} inserted at index {index}')
        else:
            print(f'Duplicate key is not allowed')

    def search(self,key):
        index=self.hashfunction(key)
        if key in self.table[index]:
            return f"{key} found at index {index}"
        else:
            return f"{key} not found"
        

    def delete(self,key):
        index=self.hashfunction(key)
        if key in self.table[index]:
            self.table[index].remove(key)
            return f"{key} removed successfully"
        else:
            return f"{key} not found"
        
    def display(self):
        print("\nHash Table")
        for i in range(self.size):
            print(f"{i}-->{self.table[i]}")
obj=HashTable(10)
obj.insert(27)
obj.insert(37)
obj.insert(47)
obj.insert(18)
obj.insert(25)
obj.insert(35)
obj.insert(15)
obj.insert(55)


# Display
obj.display()

# Search
print(obj.search(37))
print(obj.search(99))

# Delete
print(obj.delete(37))
print(obj.delete(100))

# Display Again
obj.display()
```

```python
class HashTable:
    def __init__(self,size):
        self.size=size
        self.table=[None]*size

    def hashfunction(self,key):
        return key%self.size
    
    def insert(self,key):
        index=self.hashfunction(key)
        original_index=index

        while self.table[index] is not None:
            if self.table[index]==key:
                return f"Duplicate is not allowed"
                
            index=(index+1)%self.size

            if index==original_index:
                print("Hashtable overflow")
                return 
        self.table[index]=key
        print(f"{key} inserted at index {index}")

    def search(self,key):
        index=self.hashfunction(key)
        original_index=index

        while self.table[index] is not None:
            if self.table[index]==key:
                return f"{key} found at index {index}"
            index=(index+1)%self.size

            if index==original_index:
                break
        return f"{key} not found"

    def delete(self,key):
        index=self.hashfunction(key)
        original_index=index

        while self.table[index] is not None:
            if self.table[index]==key:
                self.table[index]=None
                return f"{key} removed successfully"
            
            index=(index+1)%self.size

            if index==original_index:
                break
        return f"{key} not found"
    
    def display(self):
        print("\nHash Table")
        for i in range(self.size):
            print(f"{i}-->{self.table[i]}")

obj=HashTable(10)
obj.insert(27)
obj.insert(37)
obj.insert(47)
obj.insert(18)
obj.insert(25)
obj.insert(35)
obj.insert(15)
obj.insert(55)


# Display
obj.display()

# Search
print(obj.search(37))
print(obj.search(99))

# Delete
print(obj.delete(37))
print(obj.delete(100))

# Display Again
obj.display()

```
