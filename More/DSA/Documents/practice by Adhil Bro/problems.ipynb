```python
# X^n
class solution:
    def FindPower(self,x:int,n:int)->int:
        if n==0:
            return 1
        a=self.FindPower(x,n//2)
        if n%2==0:
            return a*a
        else:
            return a*a*x

#For negatve powers, eg: 2**(-3)= 1/(2**3)
    def MyPow(self,x:int,n:int)->int:
        if n>=0:
            return self.FindPower(x,n)
        else:
            return 1/self.MyPow(x,n*(-1))

obj=solution()
result=obj.FindPower(2,5)
print(f'Result is:{result}')
```

```python
#SPIRAL MATRIX
class solution:
    def SpiralMatrix(self,matrix:list[list[int]])->list[int]:
        ans=[]
        n=len(matrix)
        m=len(matrix[0])
        total=m*n

        colstart=0
        rowstart=0
        colend=m-1
        rowend=n-1

        count=0

        while total>count:
            #rowstart --> colstart->colnd
            for i in range(colstart,colend+1):
                ans.append(matrix[rowstart][i])
                count+=1
            rowstart+=1

            if count==total:
                break

            
            #colend--> rowstart->rowend

            for i in range(rowstart,rowend+1):
                ans.append(matrix[i][colend])
                count+=1
            colend-=1

            if count==total:
                break

            #rowend--> colend->colstart
            for i in range(colend,colstart-1,-1):
                ans.append(matrix[rowend][i])
                count+=1
            rowend-=1

            if count==total:
                break

            
            #colstart--> rowend->rowstart
            for i in range(rowend,rowstart-1,-1):
                ans.append(matrix[i][colstart])
                count+=1
            colstart+=1

            if count==total:
                break


        return ans
obj=solution()
result=obj.SpiralMatrix([[1,2,3],[4,5,6],[7,8,9]])
print(f'Result is:{result}')




```

```python
class solution:
    def TwoSum(self,nums:list[int],target:int)->list[int]:
        n=len(nums)
        dict1={}

        for i in range(n):
            rem=target-nums[i]

            if rem in dict1:
                return [dict1[rem],i]
            dict1[nums[i]]=i
obj=solution()
result=obj.TwoSum([5,1,8,2,6,4,9],12)
print(f'Result is:{result}')

print()
#ALTERNATE METHOD BY SORTING ARRAY
class solution:
    def AltTwoSum(self,nums:list[int],target:int)->list[int]:
        nums.sort()
        i=0
        j=len(nums)-1
        
        while i<j:
            sum1=nums[i]+nums[j]
            if sum1==target:
                return [i,j]
            if sum1>target:
                j-=1
            else:
                i+=1
obj=solution()
result=obj.AltTwoSum([5,1,8,2,6,4,9],12)
print(f'Result is:{result}')
```

```python
#Find First and last position of element in a sorted array
#SEARCHING RANGE

    #Lower Bound
class solution:
    def LowerBound(self,nums:list[int],target:int)->int:
        l=0
        r=len(nums)-1
        ans=len(nums)

        while l<=r:
            mid=(l+r)//2

            if nums[mid]>=target:
                ans=mid
                r=mid-1
            else:
                l=mid+1
        return ans
    
#Upper Bound
    def UpperBound(self,nums:list[int],target:int)->int:
        l=0
        r=len(nums)-1
        ans=len(nums)

        while l<=r:
            mid=(l+r)//2
            
            if nums[mid]>target:
                ans=mid
                r=mid-1
            else:
                l=mid+1
        return ans
    
    def SearchRange(self,nums:list[int],target:int)->list[int]:
        lb=self.LowerBound(nums,target)
        ub=self.UpperBound(nums,target)

        if lb==ub:
            return [-1,-1]
        else:
            return [lb,ub-1]

nums=[5,7,7,8,8,10]
target=8
obj=solution()
r1=obj.LowerBound(nums,target)
r2=obj.UpperBound(nums,target)
r3=obj.SearchRange(nums,target)
print(f'Result is:{r3}')
```

```python
class solution:
    def SearchInRoratedSortedArray(self,nums:list[int],target:int)->int:
        n=len(nums)
        l=0
        r=n-1

        while l<=r:
            mid=(l+r)//2

            if nums[mid]==target:
                return mid

            if nums[l]<=nums[mid]:
                if nums[l]<=target<nums[mid]:
                    r=mid-1
                else:
                    l=mid+1
            else:
                if nums[r]<=target<nums[mid]:
                    l=mid+1
                else:
                    r=mid-1
        return -1
obj=solution()
result=obj.SearchInRoratedSortedArray([4,5,6,7,0,1,2],0) 
print(f'Result is:{result}')               

```

```python
class solution:
    def SqrtX(self,x:int)->int:
        if x==0:
            return 0
        l=1
        r=x
        ans=0
        while l<=r:
            mid=(l+r)//2

            if mid**2<=x:
                ans=mid
                l=mid+1
            else:
                r=mid-1
        return  ans
obj=solution()
result=obj.SqrtX(25)
print(f'Result is:{result}')
```

```python
#Sort Array by parity
class solution:
    def SoryArrayByParity(self,nums:list[int])->list[int]:
        start=0
        for i in range(len(nums)):
            if nums[i]%2==0:
                temp=nums[i]
                nums[i]=nums[start]
                nums[start]=temp
                start+=1
        return nums
obj=solution()
result=obj.SoryArrayByParity([3,1,2,4])
print(f'Result is:{result}')
```

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class solution:
    def FindingMiddleLL(self,head):
        slow=head
        fast=head

        while fast!=None and fast.next!=None:
            slow=slow.next
            fast=fast.next.next

        return slow
    
    def ReverseLL(self,head):
        curr=head
        prev=None
        nxt=None

        while curr!=None:
            nxt=curr.next
            curr.next=prev
            prev=curr
            curr=nxt
        return prev
    
    def IsPalindrome(self,head):
        middle=self.FindingMiddleLL(head)
        second=self.ReverseLL(middle)
        first=head

        while second!=None:

            if first.data!=second.data:
                return False

            first=first.next
            second=second.next
        return True
    
obj = solution()

head = Node(1)
second = Node(2)
third = Node(3)
fourth = Node(2)
fifth = Node(1)

head.next = second
second.next = third
third.next = fourth
fourth.next = fifth

result = obj.IsPalindrome(head)

print(f'The LL is palindrome:{result}')
```

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

class solution:
    def LengthLL(self,head):
        count=0
        curr=head

        while curr!=None:
             count+=1
             curr=curr.next
        return count
    
    def Intersection(self,headA,headB):
        lenA=self.LengthLL(headA)
        lenB=self.LengthLL(headB)

        p1=headA
        p2=headB

        if lenA > lenB:
            dif=lenA-lenB

            for i in range(dif):
                p1=p1.next

        else:
            dif=lenB-lenA

            for i in range(dif):
                p2=p2.next

        while p1!=None and p2!=None:
            if p1==p2:
                return p1
            p1=p1.next
            p2=p2.next
        
        return None
obj = solution()

# -------- Shared Part --------
common1 = Node(8)
common2 = Node(9)
common3 = Node(10)

common1.next = common2
common2.next = common3

# -------- Linked List A --------
headA = Node(1)
a2 = Node(2)
a3 = Node(3)

headA.next = a2
a2.next = a3
a3.next = common1

# -------- Linked List B --------
headB = Node(5)
b2 = Node(6)

headB.next = b2
b2.next = common1

# -------- Find Intersection --------
result = obj.Intersection(headA,headB)

if result != None:
    print("Intersection Node:", result.data)
else:
    print("No Intersection")
```

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

class solution:
    def CycleLL(self,head):
        slow=head
        fast=head

        while fast!=None and fast.next!=None:
            slow=slow.next
            fast=fast.next.next

            if slow==fast:
                return True
        return False
obj=solution()

head=Node(1)
sec=Node(2)
thi=Node(3)
fou=Node(4)
five=Node(5)

head.next=sec
sec.next=thi
thi.next=fou
fou.next=five
five.next=thi

result=obj.CycleLL(head)
print(f'Result is:{result}')
```

```python
def CountSum(arr:list[int])->int:
    n=len(arr)
    #base case
    if n==0:
        return 0
    #recursive case
    return arr[0]+CountSum(arr[1:])
result=CountSum([1,2,3])
print(result)
```
