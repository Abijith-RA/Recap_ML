#### Question 1: Prime check (warm-up)
#### Write a function is_prime(n) that returns True if n is prime, else False.
#### Test: is_prime(2) → True, is_prime(1) → False, is_prime(29) → True, is_prime(100) → False


```python
def is_prime(num):

    for i in range(2,num):
        if num % i == 0:
            return False
        else:
            return True
is_prime(2)
```

Q4: Given n = 12, print its factors and how many there are.

Expected: 1 2 3 4 6 12 and 6

```python
n = 12

for i in range(1,n+1):
    if (n % i == 0):
        print(i)    
```

```python
def get_factors(n):

    factors = []

    for i in range(1,n+1):
        if n % i == 0:
            factors.append(i)

    return factors

n = 12
factors = get_factors(n)
print(*factors)
```
