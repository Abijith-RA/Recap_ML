```python
def preprossing(df : pd.DataFrame):
    
    numeric_col = df.select_dtypes(include="number").columns.tolist()
    category_col = df.select_dtypes(include=["object", "str"]).columns.tolist()
    
    null_in_number = df[numeric_col].isnull().sum()
    print("number colums\n\n\n", null_in_number)
    df[numeric_col] = df[numeric_col].fillna(df[numeric_col].median())
    
    print(df[numeric_col])
    
    return df , category_col
```
