```python
def missingValueHandling(data, column):
    
    if(data[column].dtype in ["object", "str"]):
        
        print(data[column].mode())
        data[column] = data[column].fillna(data[column].mode()[0])
    else:
        
        print(data[column].median())
        data[column] = data[column].fillna(data[column].median())
        
    return data[column]      
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

cols =["age", "experience_years", "salary"]
cacols = ["department", "rating"]

pipe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", MinMaxScaler())
])

df[cols] = pipe.fit_transform(df[cols])

cate_pipe = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("encoding", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("cat", cate_pipe, cacols)
],remainder="passthrough")
```
